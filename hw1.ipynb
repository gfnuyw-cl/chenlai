## 问题 1

根据下列递归公式, 计算下列 $T(n)$ 对应的的渐近上界。要求所求的边界尽可能的紧（tight）, 请写明步骤。

1. $T(1)=1; T(n)=T(n/4)+1$ for $n>1$
2. $T(1)=1;T(n)=3T(n/3)+n^2$ for $n>1$
3. $T(1)=1;T(n)=T(2n/3)+1$ for $n>1$
4. $T(1)=1;T(n)=5T(n/4)+n$ for $n>1$
5. $T(n)=1 \ for\ n \le 2 ; T(n)=T(\sqrt{n})+1 \ for \ n>2$

````python
# ## 1. T(1)=1，T(n)=T(n/4)+1

# 这是主定理形式：T(n)=aT(n/b)+f(n)
# 其中：a=1，b=4，f(n)=1
# 计算主定理中的关键项：n^(log_b a)=n^(log_4 1)=n^0=1
# 因为 f(n)=1=Θ(1)，所以属于主定理第二种情况。
# 因此：T(n)=Θ(log n)
# 答案：T(n)=Θ(log n)

# ## 2. T(1)=1，T(n)=3T(n/3)+n²

# 这是主定理形式：T(n)=aT(n/b)+f(n)
# 其中：a=3，b=3，f(n)=n²
# 计算：n^(log_b a)=n^(log_3 3)=n
# 比较 f(n) 和 n^(log_b a)：f(n)=n²，n² 比 n 增长得更快，因此递归外的代价占主导。
# 因此：T(n)=Θ(n²)
# 答案：T(n)=Θ(n²)

# ## 3. T(1)=1，T(n)=T(2n/3)+1

# 先把递归式改写为主定理形式：T(n)=T(n/(3/2))+1
# 所以：a=1，b=3/2，f(n)=1
# 计算：n^(log_b a)=n^(log_(3/2) 1)=n^0=1
# 因为 f(n)=1=Θ(1)，所以属于主定理第二种情况。
# 因此：T(n)=Θ(log n)
# 答案：T(n)=Θ(log n)

# ## 4. T(1)=1，T(n)=5T(n/4)+n

# 这是主定理形式：T(n)=aT(n/b)+f(n)
# 其中：a=5，b=4，f(n)=n
# 计算：n^(log_b a)=n^(log_4 5)
# 因为 log_4 5>1，所以 n^(log_4 5)>n，也就是说递归部分的代价比递归外的代价更大，递归部分占主导。
# 因此：T(n)=Θ(n^(log_4 5))
# 其中 log_4 5≈1.16，所以也可以写成 T(n)=Θ(n^1.16)
# 答案：T(n)=Θ(n^(log_4 5))

# ## 5. T(n)=1，n≤2；T(n)=T(√n)+1，n>2

# 这个递归式不能直接使用普通主定理，因为子问题规模不是 n/b，而是 √n。
# 对递归式进行展开：T(n)=T(n^(1/2))+1
# 继续展开：T(n)=T(n^(1/4))+2
# 再继续展开：T(n)=T(n^(1/8))+3
# 所以展开 k 次后：T(n)=T(n^(1/2^k))+k
# 递归停止条件为：n^(1/2^k)≤2
# 两边取对数：log n / 2^k ≤ 1
# 所以：2^k ≥ log n
# 再取一次对数：k ≥ log log n
# 因此递归层数为：Θ(log log n)
# 所以：T(n)=Θ(log log n)
# 答案：T(n)=Θ(log log n)
````

idea:

````python
# 1. 时间复杂度递归式：
# 设 T(n) 表示处理长度为 n 的数组所需时间。
# 当 n <= 2 时，只进行常数次比较和交换，所以 T(n) = Θ(1)。
# 当 n >= 3 时，算法递归调用 3 次，每次规模为 ceil(2n/3)，递归外只做常数次操作。
# 因此递归式为：
# T(n) = 3T(ceil(2n/3)) + Θ(1)
# 忽略取整对渐近复杂度的影响：
# T(n) = 3T(2n/3) + Θ(1)
# 写成主定理形式：
# T(n) = aT(n/b) + f(n)
# 其中 a = 3，b = 3/2，f(n) = Θ(1)
# 计算主定理关键项：
# n^(log_b a) = n^(log_(3/2)3)
# 因为 f(n) = Θ(1) 比 n^(log_(3/2)3) 小，所以递归部分占主导。
# 所以：
# T(n) = Θ(n^(log_(3/2)3)) ≈ Θ(n^2.71)


# 2. 算法功能：
# 该算法的功能是将数组 A[p..r] 按升序排序。
# 它本质上是 Stooge Sort，也叫臭皮匠排序。
# 算法思想是：
# 先排序前 2/3 区间，再排序后 2/3 区间，最后再次排序前 2/3 区间。
# 该算法不是最高效的解决方案。
# 因为它的时间复杂度约为 Θ(n^2.71)，比归并排序、堆排序、快速排序平均情况的 O(nlogn) 更慢。
# 因此，该算法可以完成排序，但效率较低，不是最高效的排序算法。


# 3. Python 实现该算法或其对应的更高效版本的算法
# 这里实现更高效版本：归并排序
# 归并排序的时间复杂度是 O(nlogn)，比题目中的 Stooge Sort 更高效。

def merge_sort(A):
    if len(A) <= 1:
        return A

    mid = len(A) // 2

    left = merge_sort(A[:mid])
    right = merge_sort(A[mid:])

    return merge(left, right)


def merge(left, right):
    result = []
    i = 0
    j = 0

    while i < len(left) and j < len(right):
        if left[i] <= right[j]:
            result.append(left[i])
            i += 1
        else:
            result.append(right[j])
            j += 1

    while i < len(left):
        result.append(left[i])
        i += 1

    while j < len(right):
        result.append(right[j])
        j += 1

    return result


A = [5, 2, 4, 7, 1, 3, 2, 6]

print("排序前：", A)

A = merge_sort(A)

print("排序后：", A)

# your algorithm time complexity is: O(nlogn)
````

## 问题 3

给定一个包含n个元素的数组 `profits` , 它的第 `i` 个元素 `profits[i]` 表示一支股票第 `i` 天的**收益**（正数表示涨, 负数表示跌）。你只能选择 **某一天** 买入这只股票, 并选择在 **未来的某一个不同的日子** 卖出该股票。

1. 设计一个算法来计算你所能获取的最大利润和对应买入和卖出的日期。请分析算法方案, 计算其时间复杂度, 并且使用python编程实现该算法。

2. \* 设计一个时间复杂度为 $O(n)$的算法实现该算法

e.g. :
---
profits=[3,2,1,-7,5,2,-1,3,-1], 第5天买入, 第8天卖出, 收益最大：9

idea:

````python
# add your code here
# 问题 3

# 1. 算法思想：
# profits[i] 表示第 i 天的收益，正数表示涨，负数表示跌。
# 如果从第 l 天买入，第 r 天卖出，那么总收益就是：
# profits[l] + profits[l+1] + ... + profits[r]
# 所以这个问题可以转化为“最大子数组问题”。
#
# 也就是：在 profits 数组中，找到一个连续子数组，使它的和最大。
# 这个连续子数组的起点就是买入日期，终点就是卖出日期。
#
# 可以使用 Kadane 算法：
# current_sum 表示以当前天结尾的最大连续收益。
# max_sum 表示目前找到的最大收益。
# 如果 current_sum 小于 0，说明前面的收益会拖累后面的结果，
# 所以从当前天重新开始计算。
#
# 2. 时间复杂度：
# 该算法只需要从左到右遍历数组一次，所以时间复杂度是 O(n)。
# 只使用了几个变量，所以空间复杂度是 O(1)。


# 3. Python 实现：
def max_profit(profits):
    n = len(profits)

    current_sum = profits[0]
    max_sum = profits[0]

    start = 0
    best_start = 0
    best_end = 0

    for i in range(1, n):
        # 如果前面的连续收益小于 0，就从当前天重新开始
        if current_sum < 0:
            current_sum = profits[i]
            start = i
        else:
            current_sum += profits[i]

        # 更新最大收益和对应的买入、卖出日期
        if current_sum > max_sum:
            max_sum = current_sum
            best_start = start
            best_end = i

    # 题目中的日期从第 1 天开始，所以数组下标要加 1
    buy_day = best_start + 1
    sell_day = best_end + 1

    return max_sum, buy_day, sell_day


profits = [3, 2, 1, -7, 5, 2, -1, 3, -1]

profit, buy_day, sell_day = max_profit(profits)

print("最大收益:", profit)
print("买入日期: 第", buy_day, "天")
print("卖出日期: 第", sell_day, "天")

# your algorithm time complexity is: O(n)
````

## 问题 4

给定一个包含不同整数元素的数组 $ A[1..n]$ ,并且满足条件：$A[1]>A[2]$ 并且 $A[n-1]<A[n]$; 规定：如果一个元素比它两边的邻居元素都小, 即：$A[x]<A[x-1], A[x]<A[x+1]$ , 称这个元素A[x]为“局部最小”。通过遍历一次数组, 我们可以很容易在 $O(n)$的时间复杂度下找到一个局部最小值, 


1. 分析该问题, 设计一个算法在$O(logn)$的时间复杂度下找到一个局部最小(返回数值), 要求：分析算法设计思路, 并且使用python编程实现
2. \* 设计算法找出所有局部最小值, 分析算法设计思路, 并使用python编程实现

e.g.:
---
A=[9, 3, 7, 2, 1, 4, 5 ] 时,  局部最小元素为 3, 1

idea:

````python
# add your code here
# 问题 4

# 1. 算法思想：
# 题目要求在数组 A 中找到一个局部最小值。
# 局部最小值定义为：
# A[x] < A[x-1] 且 A[x] < A[x+1]
#
# 已知条件：
# A[1] > A[2]
# A[n-1] < A[n]
#
# 也就是说，数组左边一开始是下降趋势，右边最后是上升趋势。
# 因此数组中间一定存在至少一个局部最小值。
#
# 可以使用二分查找思想：
# 每次取中间位置 mid。
# 如果 A[mid] 同时小于左右邻居，那么 A[mid] 就是局部最小值。
# 如果 A[mid] > A[mid-1]，说明左边方向存在下降趋势，
# 因此局部最小值一定在左半部分。
# 否则，如果 A[mid] > A[mid+1]，说明右边方向存在下降趋势，
# 因此局部最小值一定在右半部分。
#
# 每次都可以排除一半区间，所以时间复杂度是 O(logn)。
# 空间复杂度是 O(1)。


# 2. Python 实现：
def find_local_min(A):
    n = len(A)

    left = 1
    right = n - 2

    while left <= right:
        mid = (left + right) // 2

        # 如果 mid 位置比左右邻居都小，说明找到局部最小值
        if A[mid] < A[mid - 1] and A[mid] < A[mid + 1]:
            return A[mid]

        # 如果左边邻居更小，说明左半部分一定存在局部最小值
        elif A[mid] > A[mid - 1]:
            right = mid - 1

        # 否则右边邻居更小，说明右半部分一定存在局部最小值
        else:
            left = mid + 1


A = [9, 3, 7, 2, 1, 4, 5]

answer = find_local_min(A)

print("局部最小值:", answer)

# your algorithm time complexity is: O(logn)
````

给定包含n个不同数字的一组数, 寻找一种基于比较的算法在这组数中找到k个最小的数字, 并按顺序输出它们。

1. 将n个数先进行排序, 然后按顺序输出最小的k个数。要求：选择合适的排序算法实现上述操作, 计算算法时间复杂度, 并使用python编程实现。

3. \* 假设数组中包含的数据总数目超过了计算机的存储能力, 请设计一个算法, 找到这堆数据的前k小的数值, 计算时间复杂度, 并使用python实现该算法, 假设计算机一定能存储k个数据。

e.g.：
---
数组arr=[5,4,3,2,6,1,88,33,22,107] 的前3个最小数据为：1, 2, 3

idea:

````python
# add your code here
# 问题 5


# 1. 先排序，然后输出最小的 k 个数

# 算法思想：
# 先使用基于比较的排序算法，把整个数组从小到大排序。
# 排序完成后，数组前 k 个元素就是最小的 k 个数。
#
# 时间复杂度：
# 排序需要 O(nlogn)，输出前 k 个数需要 O(k)。
# 因为 k <= n，所以总时间复杂度为 O(nlogn)。
# 空间复杂度取决于排序算法，这里使用 Python 内置 sorted，空间复杂度为 O(n)。

def k_smallest_by_sort(arr, k):
    sorted_arr = sorted(arr)
    return sorted_arr[:k]


arr = [5, 4, 3, 2, 6, 1, 88, 33, 22, 107]
k = 3

print("方法一结果:", k_smallest_by_sort(arr, k))

# your algorithm time complexity is: O(nlogn)



# 2. 建立最小堆，然后调用 k 次 Extract-min

# 算法思想：
# 先建立一个最小堆。
# 题目要求使用空堆不断插入元素的方式建堆，所以从空堆开始，
# 将数组中的 n 个元素依次插入堆中。
# 最小堆的堆顶永远是当前最小值。
# 因此连续执行 k 次 Extract-min，就可以按顺序得到最小的 k 个数。
#
# 时间复杂度：
# 插入一个元素需要 O(logn)，插入 n 个元素需要 O(nlogn)。
# Extract-min 一次需要 O(logn)，执行 k 次需要 O(klogn)。
# 所以总时间复杂度为 O(nlogn + klogn)，也就是 O((n+k)logn)。
# 空间复杂度为 O(n)，因为堆中需要存储 n 个元素。

class MinHeap:
    def __init__(self):
        self.heap = []

    def insert(self, x):
        self.heap.append(x)
        self.sift_up(len(self.heap) - 1)

    def extract_min(self):
        if len(self.heap) == 0:
            return None

        min_value = self.heap[0]
        last_value = self.heap.pop()

        if len(self.heap) > 0:
            self.heap[0] = last_value
            self.sift_down(0)

        return min_value

    def sift_up(self, index):
        while index > 0:
            parent = (index - 1) // 2

            if self.heap[parent] <= self.heap[index]:
                break

            self.heap[parent], self.heap[index] = self.heap[index], self.heap[parent]
            index = parent

    def sift_down(self, index):
        n = len(self.heap)

        while True:
            left = 2 * index + 1
            right = 2 * index + 2
            smallest = index

            if left < n and self.heap[left] < self.heap[smallest]:
                smallest = left

            if right < n and self.heap[right] < self.heap[smallest]:
                smallest = right

            if smallest == index:
                break

            self.heap[index], self.heap[smallest] = self.heap[smallest], self.heap[index]
            index = smallest


def k_smallest_by_min_heap(arr, k):
    heap = MinHeap()

    for x in arr:
        heap.insert(x)

    result = []

    for i in range(k):
        result.append(heap.extract_min())

    return result


print("方法二结果:", k_smallest_by_min_heap(arr, k))

# your algorithm time complexity is: O(nlogn + klogn)



# 3. 数据量太大，不能全部存入内存时，找到最小的 k 个数

# 算法思想：
# 如果数组太大，计算机不能存储全部数据，但是一定能存储 k 个数据，
# 那么可以维护一个大小为 k 的最大堆。
#
# 最大堆中保存当前已经读到的数据里最小的 k 个数。
# 堆顶是这 k 个数中最大的那个。
#
# 遍历数据时：
# 1. 如果堆中元素少于 k 个，直接插入。
# 2. 如果当前元素 x 小于堆顶，说明 x 应该进入最小的 k 个数中，
#    此时删除堆顶，再插入 x。
# 3. 如果当前元素 x 大于等于堆顶，说明 x 不可能属于最小的 k 个数，直接跳过。
#
# 最后堆中剩下的就是最小的 k 个数。
# 为了按顺序输出，可以最后再排序一次。
#
# 时间复杂度：
# 每个元素最多进行一次堆操作，堆大小始终为 k，所以每次操作是 O(logk)。
# 一共有 n 个元素，所以总时间复杂度是 O(nlogk)。
# 最后排序 k 个数需要 O(klogk)。
# 因此总时间复杂度是 O(nlogk + klogk)，通常写作 O(nlogk)。
# 空间复杂度是 O(k)。

import heapq

def k_smallest_large_data(data, k):
    max_heap = []

    for x in data:
        if len(max_heap) < k:
            heapq.heappush(max_heap, -x)
        else:
            if x < -max_heap[0]:
                heapq.heappop(max_heap)
                heapq.heappush(max_heap, -x)

    result = []

    for x in max_heap:
        result.append(-x)

    result.sort()

    return result


print("方法三结果:", k_smallest_large_data(arr, k))

# your algorithm time complexity is: O(nlogk)
````

2. 输入：A=[23,-8, 9, -3, -7, 9, -23, 22], 输出：[23, 22]

idea:

````python
# add your code here
# 问题 6

# 算法思想：
# 使用栈 stack 来模拟星球碰撞过程。
#
# 数组中的正数表示星球从左向右运动，负数表示星球从右向左运动。
# 只有一种情况会发生碰撞：
# 栈顶星球是正数，当前星球是负数。
#
# 也就是：
# stack[-1] > 0 且 x < 0
#
# 碰撞时：
# 1. 如果栈顶星球质量小于当前星球，栈顶星球消失，当前星球继续向左运动，
#    可能继续和更前面的星球碰撞。
# 2. 如果栈顶星球质量大于当前星球，当前星球消失。
# 3. 如果两个星球质量相同，根据题目要求，右向左运动的星球消失，
#    自左向右运动的星球保持不变。
#
# 遍历完整个数组后，栈中剩下的星球就是最终存续的星球。


# 时间复杂度：
# 每个星球最多入栈一次，最多出栈一次。
# 所以总时间复杂度是 O(n)。
# 空间复杂度是 O(n)，因为最坏情况下所有星球都不会碰撞，需要全部存入栈中。


# Python 实现：
def asteroid_collision(A):
    stack = []

    for x in A:
        alive = True

        # 只有栈顶为正数，当前星球为负数时，才会发生碰撞
        while alive and stack and stack[-1] > 0 and x < 0:
            if stack[-1] < -x:
                # 栈顶星球质量小，栈顶星球消失
                stack.pop()

            elif stack[-1] > -x:
                # 当前星球质量小，当前星球消失
                alive = False

            else:
                # 质量相同，根据题意，右向左运动的星球消失
                # 当前星球 x 是负数，表示右向左运动，所以当前星球消失
                alive = False

        # 如果当前星球没有消失，就入栈
        if alive:
            stack.append(x)

    return stack


A = [23, -8, 9, -3, -7, 9, -23, 22]

result = asteroid_collision(A)

print("最终星球存续情况:", result)

# your algorithm time complexity is: O(n)
````

## 问题 7

> 0-1背包问题：现在有4块大理石，每个大理石对应的重量和价值使用一个元组表示, 即（重量，价值），4块大理石的重量和价值为：[(5,10), (4,40), (6,30), (3, 50)], 假设你有一辆最大承重为9的小推车，请问使用这个小推车装大理石的最大价值为多少

请设计一个算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

````python
# add your code here
# 问题 7

# 算法思想：
# 这是一个 0-1 背包问题。
# 每块大理石只能选择一次，要么装入小推车，要么不装入。
#
# 使用动态规划解决。
# 定义 dp[i][w] 表示：
# 只考虑前 i 块大理石，在小推车最大承重为 w 时，可以获得的最大价值。
#
# 对于第 i 块大理石，有两种选择：
# 1. 不装入第 i 块大理石：
#    dp[i][w] = dp[i-1][w]
#
# 2. 如果当前容量 w 能装下第 i 块大理石，就可以选择装入：
#    dp[i][w] = dp[i-1][w-weight] + value
#
# 因此状态转移方程为：
# dp[i][w] = max(dp[i-1][w], dp[i-1][w-weight] + value)
#
# 最终答案是 dp[n][capacity]。


# 时间复杂度：
# 设 n 表示大理石数量，W 表示小推车最大承重。
# 动态规划需要遍历 n 个大理石，并且对每个大理石遍历 0 到 W 的所有容量。
# 所以时间复杂度是 O(nW)。
# 空间复杂度是 O(nW)，因为使用了二维 dp 数组。


# Python 实现：
def knapsack_01(items, capacity):
    n = len(items)

    # dp[i][w] 表示前 i 个大理石，在容量为 w 时能获得的最大价值
    dp = [[0] * (capacity + 1) for _ in range(n + 1)]

    for i in range(1, n + 1):
        weight, value = items[i - 1]

        for w in range(capacity + 1):
            # 情况 1：不选择第 i 个大理石
            dp[i][w] = dp[i - 1][w]

            # 情况 2：如果能装下第 i 个大理石，就尝试选择它
            if w >= weight:
                dp[i][w] = max(dp[i][w], dp[i - 1][w - weight] + value)

    return dp[n][capacity]


items = [(5, 10), (4, 40), (6, 30), (3, 50)]
capacity = 9

answer = knapsack_01(items, capacity)

print("最大价值:", answer)

# your algorithm time complexity is: O(nW)|
````

给你二叉搜索树的根节点 root ，该树中的恰好两个节点的值被错误地交换。请在不改变其结构的情况下，恢复这棵树 。设计算法解决该问题，分析算法设计思路，计算时间复杂度，并基于python编程实现

e.g.:
---
1.

<div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入：root = [1,3,null,null,2]， 输出为：[3,1,null,null,2]， 

解释：3 不能是 1 的左孩子，因为 3 > 1 。交换 1 和 3 使二叉搜索树有效。

 <div style="text-align:center;width:200px;"> <img alt="" src="./fig/hw3q9-1.jpg"> </div>
输入： root=[3,1,4,null,null,2]， 输出为：[2,1,4,null,null,3]

解释：2 不能在 3 的右子树中，因为 2 < 3 。交换 2 和 3 使二叉搜索树有效。

idea:

````python
# add your code here
# 问题 8

# 算法思想：
# 二叉搜索树的中序遍历结果应该是升序序列。
# 如果有两个节点的值被错误交换，那么中序遍历时一定会出现逆序情况。
#
# 例如正常中序序列：
# 1, 2, 3, 4
#
# 如果 2 和 4 被交换：
# 1, 4, 3, 2
#
# 遍历时如果发现 prev.val > current.val，说明这两个节点的位置有问题。
# 第一次发现逆序时，记录 first = prev，second = current。
# 如果第二次发现逆序，更新 second = current。
# 最后交换 first 和 second 的值即可恢复二叉搜索树。
#
# 注意：
# 题目要求不能改变树的结构，所以只能交换两个节点的 val。


# 时间复杂度：
# 中序遍历需要访问每个节点一次，所以时间复杂度是 O(n)。
# n 表示二叉搜索树中的节点数量。
# 空间复杂度是 O(h)，h 表示树的高度，来自递归调用栈。


# Python 实现：
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def recover_bst(root):
    first = None
    second = None
    prev = None

    def inorder(node):
        nonlocal first, second, prev

        if node is None:
            return

        inorder(node.left)

        # 如果前一个节点的值大于当前节点的值，说明出现逆序
        if prev is not None and prev.val > node.val:
            if first is None:
                first = prev
            second = node

        prev = node

        inorder(node.right)

    inorder(root)

    # 只交换两个错误节点的值，不改变树的结构
    if first is not None and second is not None:
        first.val, second.val = second.val, first.val

    return root


# 辅助函数：用层序数组建立二叉树，None 表示空节点
def build_tree(arr):
    if not arr:
        return None

    nodes = []

    for x in arr:
        if x is None:
            nodes.append(None)
        else:
            nodes.append(TreeNode(x))

    j = 1

    for i in range(len(nodes)):
        if nodes[i] is not None:
            if j < len(nodes):
                nodes[i].left = nodes[j]
                j += 1

            if j < len(nodes):
                nodes[i].right = nodes[j]
                j += 1

    return nodes[0]


# 辅助函数：将二叉树转换成层序数组
def tree_to_list(root):
    if root is None:
        return []

    queue = [root]
    result = []

    while queue:
        node = queue.pop(0)

        if node is None:
            result.append(None)
        else:
            result.append(node.val)
            queue.append(node.left)
            queue.append(node.right)

    # 删除末尾多余的 None
    while result and result[-1] is None:
        result.pop()

    return result


# test 1
root1 = build_tree([1, 3, None, None, 2])
recover_bst(root1)
print(tree_to_list(root1))


# test 2
root2 = build_tree([3, 1, 4, None, None, 2])
recover_bst(root2)
print(tree_to_list(root2))

# your algorithm time complexity is: O(n)
````

给定 `n` 个非负整数表示每个宽度为 `1` 的柱子的高度图，计算按此排列的柱子，下雨之后能接多少雨水。

请分析算法设计思路，计算时间复杂度，并基于python编程实现

idea:

````python
# add your code here
# 问题 9

# 算法思想：
# 这是经典的接雨水问题。
# 对于每一个位置 i，它能接多少水，取决于它左边最高柱子和右边最高柱子中的较小值。
#
# 也就是说：
# water[i] = min(left_max[i], right_max[i]) - height[i]
#
# 如果这个值大于 0，说明当前位置可以接水；
# 如果小于等于 0，说明当前位置不能接水。
#
# 可以先从左到右计算每个位置左边的最大高度 left_max。
# 再从右到左计算每个位置右边的最大高度 right_max。
# 最后遍历每个位置，累加能接的雨水量。


# 时间复杂度：
# 需要遍历数组三次，所以时间复杂度是 O(n)。
# 使用了两个辅助数组 left_max 和 right_max，所以空间复杂度是 O(n)。


# Python 实现：
def trap_rain_water(height):
    n = len(height)

    if n == 0:
        return 0

    left_max = [0] * n
    right_max = [0] * n

    # left_max[i] 表示位置 i 左边包括自己在内的最高柱子高度
    left_max[0] = height[0]
    for i in range(1, n):
        left_max[i] = max(left_max[i - 1], height[i])

    # right_max[i] 表示位置 i 右边包括自己在内的最高柱子高度
    right_max[n - 1] = height[n - 1]
    for i in range(n - 2, -1, -1):
        right_max[i] = max(right_max[i + 1], height[i])

    water = 0

    # 每个位置能接的水量等于左右最高柱子的较小值减去当前柱子高度
    for i in range(n):
        water += min(left_max[i], right_max[i]) - height[i]

    return water


# test 1
height1 = [0, 1, 0, 2, 1, 0, 1, 3, 2, 1, 2, 1]
print(trap_rain_water(height1))


# test 2
height2 = [4, 2, 0, 3, 2, 5]
print(trap_rain_water(height2))

# your algorithm time complexity is: O(n)
````

## 问题 10

给你一个二叉树的根节点 root ，树中每个节点都存放有一个 0 到 9 之间的数字。每条从根节点到叶节点的路径都代表一个数字：

设计一个算法计算从根节点到叶节点生成的所有数字之和 ，分析算法设计思路，计算时间复杂度，并基于python编程实现

叶节点:是指没有子节点的节点。

e.g.:

 输入：root = [1,2,3], 输出：25

 输入：root = [4,9,0,5,1], 输出：1026

解释：从根到叶子节点路径 4->9->5 代表数字 495，4->9->1 代表数字 491，4->0 代表数字 40。因此和为 495 + 491 + 40 = 1026 。

idea:

````python
# add your code here
# 问题 10

# 算法思想：
# 使用深度优先搜索 DFS。
# 从根节点到叶子节点的每一条路径都可以组成一个数字。
#
# 例如路径 1 -> 2 -> 3：
# 先得到 1
# 再得到 1 * 10 + 2 = 12
# 再得到 12 * 10 + 3 = 123
#
# 所以在 DFS 过程中，需要维护当前路径已经形成的数字 current。
# 每访问一个新节点 node，就更新：
# current = current * 10 + node.val
#
# 如果当前节点是叶子节点，说明已经形成了一个完整数字，
# 直接返回 current。
#
# 如果当前节点不是叶子节点，就继续递归计算左子树和右子树。
# 最终答案就是所有根节点到叶子节点路径形成数字的总和。


# 时间复杂度：
# 每个节点只会被访问一次，所以时间复杂度是 O(n)。
# n 表示二叉树中的节点数量。
# 空间复杂度是 O(h)，h 表示树的高度，来自递归调用栈。


# Python 实现：
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right


def sum_root_to_leaf_numbers(root):
    def dfs(node, current):
        if node is None:
            return 0

        # 更新当前路径形成的数字
        current = current * 10 + node.val

        # 如果是叶子节点，返回当前完整数字
        if node.left is None and node.right is None:
            return current

        # 递归计算左右子树路径数字之和
        return dfs(node.left, current) + dfs(node.right, current)

    return dfs(root, 0)


# 辅助函数：用层序数组建立二叉树，None 表示空节点
def build_tree(arr):
    if not arr:
        return None

    nodes = []

    for x in arr:
        if x is None:
            nodes.append(None)
        else:
            nodes.append(TreeNode(x))

    j = 1

    for i in range(len(nodes)):
        if nodes[i] is not None:
            if j < len(nodes):
                nodes[i].left = nodes[j]
                j += 1

            if j < len(nodes):
                nodes[i].right = nodes[j]
                j += 1

    return nodes[0]


# test 1
root1 = build_tree([1, 2, 3])
print(sum_root_to_leaf_numbers(root1))


# test 2
root2 = build_tree([4, 9, 0, 5, 1])
print(sum_root_to_leaf_numbers(root2))

# your algorithm time complexity is: O(n)
````
